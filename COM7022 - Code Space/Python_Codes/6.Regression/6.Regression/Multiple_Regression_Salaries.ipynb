```python
import pandas
```

```python
df = pandas.read_csv("Salaries.csv")
```

```python
print(df)
```

```python
d_education = {'Bs': 0, 'Ms':1, 'Phd':2}
df['Education'] = df['Education'].map(d_education)

print(df)
```

```python
features = ['Age','Experience','Education']

X = df[features]
y = df['Salary']
```

```python
print(X)
print(y)
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(10)

X_poly = poly.fit_transform(X)

print(X_poly)
```

```python
from sklearn import linear_model
```

```python
regr = linear_model.LinearRegression()
```

```python
# For Polynomial Regression
regr.fit(X_poly, y)
```

```python
# For Linear Regression
regr.fit(X, y)
```

```python
# For Polynomial Regression
predicted_output = regr.predict(X_poly)
```

```python
# For Linear Regression
predicted_output = regr.predict(X)
```

```python
print(predicted_output)
```
