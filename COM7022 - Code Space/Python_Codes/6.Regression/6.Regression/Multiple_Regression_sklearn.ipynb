```python
import pandas
```

```python
df = pandas.read_csv("Cars.csv")
```

```python
print(df)
```

```python
from sklearn import linear_model
```

```python
X = df[["Weight","Volume"]]
y = df["CO2"]
```

```python
regr = linear_model.LinearRegression()
regr.fit(X, y)
```

```python
predictedCO2 = regr.predict(X)
print(predictedCO2)
```
