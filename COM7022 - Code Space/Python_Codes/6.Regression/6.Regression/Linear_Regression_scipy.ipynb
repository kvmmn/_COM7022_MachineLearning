```python
import matplotlib.pyplot as plt
```

```python
x = [5,7,8,7,2,17,2,9,4,11,12,9,6]
y = [99,86,87,88,111,86,103,87,94,78,77,85,86]

plt.scatter(x, y)
plt.show()
```

```python
from scipy import stats
```

```python
slope, intercept, r, p, std_err = stats.linregress(x, y)
```

```python
print(slope, intercept, r, p, std_err)
```

```python
def linear_func(x):
  return slope * x + intercept
```

```python
output = list(map(linear_func, x))
```

```python
output = []

for i in x:
  out = linear_func(i)
  output.append(out)
```

```python
print(output)
print(y)
```

```python
plt.scatter(x, y)
plt.plot(x, output)
plt.show()
```
