```python
import numpy as np

# Input data (XOR example)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])  # 4 examples, 2 features each
y = np.array([[0], [1], [1], [0]])  # XOR outputs

# Initialize weights and biases
input_size = X.shape[1]  # 2 features
hidden_size = 2  # 2 neurons in the hidden layer
output_size = 1  # 1 output neuron

# Weights and biases initialization
np.random.seed(42)
weights_input_hidden = np.random.rand(input_size, hidden_size)  # (2, 2) weights for input to hidden
bias_hidden = np.random.rand(1, hidden_size)  # (1, 2) bias for hidden layer

weights_hidden_output = np.random.rand(hidden_size, output_size)  # (2, 1) weights for hidden to output
bias_output = np.random.rand(1, output_size)  # (1, 1) bias for output layer

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Sigmoid derivative (for backpropagation)
def sigmoid_derivative(x):
    return x * (1 - x)

# Training the network
learning_rate = 0.1
epochs = 10000

# Forward pass and Backpropagation
for epoch in range(epochs):
    # Forward pass
    # Input to hidden layer
    hidden_input = np.dot(X, weights_input_hidden) + bias_hidden
    hidden_output = sigmoid(hidden_input)  # Output of hidden layer

    # Hidden to output layer
    output_input = np.dot(hidden_output, weights_hidden_output) + bias_output
    predicted_output = sigmoid(output_input)  # Final output

    # Calculate the error (difference between predicted and actual)
    error = y - predicted_output

    # Backpropagation
    # Output layer: Calculate gradient
    output_error_term = error * sigmoid_derivative(predicted_output)

    # Hidden layer: Calculate gradient
    hidden_error_term = output_error_term.dot(weights_hidden_output.T) * sigmoid_derivative(hidden_output)

    # Update weights and biases using gradient descent
    weights_hidden_output += hidden_output.T.dot(output_error_term) * learning_rate
    bias_output += np.sum(output_error_term, axis=0, keepdims=True) * learning_rate

    weights_input_hidden += X.T.dot(hidden_error_term) * learning_rate
    bias_hidden += np.sum(hidden_error_term, axis=0, keepdims=True) * learning_rate

    # Optionally print the error at every 1000 epochs
    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Error: {np.mean(np.abs(error))}")

# After training, print the final output
print("\nFinal Output after training:")
print(predicted_output)
```

**Explanation of Code:**
Data (X and y):
The input matrix X consists of 4 training samples, each with 2 features. These are the inputs for the XOR problem.
The target output y is the XOR result for each input combination.

Weights and Biases:
Weights and biases are initialized randomly for both layers:
weights_input_hidden: A weight matrix between the input and hidden layer.
bias_hidden: Bias vector for the hidden layer.
weights_hidden_output: A weight matrix between the hidden layer and output layer.

bias_output: Bias for the output layer.

Forward Pass:
The input is passed through the network:
The input X is multiplied by the weights of the input-to-hidden layer and the bias is added. The result is passed through the sigmoid activation to get the hidden layer's output.
The hidden layer output is then multiplied by the weights of the hidden-to-output layer, and the output bias is added. This gives the final output which is passed through the sigmoid activation.

Backpropagation:
The error (difference between predicted output and actual output) is calculated.
Using gradient descent, we calculate the gradients of the error with respect to the weights and biases. This is done in two parts:
Output layer error term: The error term for the output layer is computed using the derivative of the sigmoid function.
Hidden layer error term: The error term for the hidden layer is computed by propagating the error back through the hidden-to-output weights.
The weights and biases are updated using these error terms, and this process is repeated for the specified number of epochs (iterations).


Training:
The network is trained for 10000 epochs, where at each epoch, the forward pass and backpropagation steps occur.
The mean error is printed every 1000 epochs to monitor training progress.


Output:
After training, the neural network should be able to predict the XOR function correctly.
