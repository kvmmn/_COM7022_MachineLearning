```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score
```

```python
# Generate sample data
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
```

```python
# Visualize the data
plt.figure(figsize=(8, 8))
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Data Before Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
# Apply K-Means Clustering
kmeans = KMeans(n_clusters=4, random_state=0)
kmeans.fit(X)
```

```python
# Get cluster centers and labels
centers = kmeans.cluster_centers_
labels = kmeans.labels_
```

```python
# Visualize the Clusters
plt.figure(figsize=(8, 8))
plt.scatter(X[:, 0], X[:, 1], c=labels, s=50, cmap='viridis')  # Clustered data
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centers')  # Centers
plt.title("Clusters with K-Means")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

```python
# Calculate the Silhouette Score
silhouette_avg = silhouette_score(X, labels)
print(f"Silhouette Score for 4 clusters: {silhouette_avg:.2f}")
```
