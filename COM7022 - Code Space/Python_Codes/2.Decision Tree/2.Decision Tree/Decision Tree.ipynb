```python
import pandas
```

```python
df = pandas.read_csv("Customers.csv")

print(df)
```

```python
print(df['Buy_computer'])
```

```python
d = {'young': 0, 'middle': 1, 'old': 2}
df['Age'] = df['Age'].map(d)
print(df)
```

```python
d = {'high': 0, 'medium': 1, 'low': 2}
df['Income'] = df['Income'].map(d)

d = {'no': 0, 'yes': 1}
df['Student'] = df['Student'].map(d)

d = {'fair': 0, 'excellent': 1}
df['Credit_rating'] = df['Credit_rating'].map(d)

d = {'no': 0, 'yes': 1}
df['Buy_computer'] = df['Buy_computer'].map(d)

print(df)
```

```python
features = ['Age', 'Income', 'Student', 'Credit_rating']

X = df[features]
y = df['Buy_computer']

print(X)
print('\n\n')
print(y)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree = DecisionTreeClassifier(criterion='entropy')
dtree = dtree.fit(X, y)
```

```python
test = pandas.read_csv("test_set.csv")

d = {'young': 0, 'middle': 1, 'old': 2}
test['Age'] = test['Age'].map(d)

d = {'high': 0, 'medium': 1, 'low': 2}
test['Income'] = test['Income'].map(d)

d = {'no': 0, 'yes': 1}
test['Student'] = test['Student'].map(d)

d = {'fair': 0, 'excellent': 1}
test['Credit_rating'] = test['Credit_rating'].map(d)

print(test)
```

```python
y_pred = dtree.predict(test)
print(y_pred)
```

```python
from sklearn import tree
import pydotplus
import matplotlib.pyplot as plt
import matplotlib.image as pltimg
```

```python
data = tree.export_graphviz(dtree, out_file=None, feature_names=features)

graph = pydotplus.graph_from_dot_data(data)
graph.write_png('mydecisiontree.png')

img=pltimg.imread('mydecisiontree.png')
imgplot = plt.imshow(img)
plt.show()
```
