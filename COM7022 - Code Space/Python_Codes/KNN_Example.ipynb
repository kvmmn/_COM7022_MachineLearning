```python
import math
from collections import Counter
```

```python
data = [
    {"maths": 4, "cs": 3, "result": "fail"},
    {"maths": 6, "cs": 7, "result": "pass"},
    {"maths": 7, "cs": 8, "result": "pass"},
    {"maths": 5, "cs": 5, "result": "fail"},
    {"maths": 8, "cs": 8, "result": "pass"},
]
```

```python
# Query point
query = {"maths": 6, "cs": 8}
k = 3
```

```python
# Function to calculate Euclidean distance
def euclidean_distance(p1, p2):
    return math.sqrt((p1["maths"] - p2["maths"]) ** 2 + (p1["cs"] - p2["cs"]) ** 2)
```

```python
# Compute distances
distances = []
for point in data:
    dist = euclidean_distance(query, point)
    distances.append((dist, point["result"]))
```

```python
# Sort distances and get the nearest k neighbors
distances.sort(key=lambda x: x[0])
k_neighbors = distances[:k]
```

```python

# Get the majority vote
results = [neighbor[1] for neighbor in k_neighbors]
most_common = Counter(results).most_common(1)[0][0]
```

```python
print(f"The predicted result for the query {query} is: {most_common}")
```
