```python
def min(list):
  lowest = list[0]
  for n in list:
    if n < lowest:
      lowest = n

  return lowest
```

```python
def max(list):
  greatest = list[0]
  for n in list:
    if n > greatest:
      greatest = n

  return greatest
```

```python
def sum(list):
  total = 0
  for n in list:
    total = total + n
    
  return total
```

```python
def average(list):
  avg = sum(list)/len(list)
  return avg
```

```python
def median(list):
  list.sort()
  length = len(list)

  if(length % 2 == 1):
    middle = (int) (length / 2)
    return list[middle]
  
  if(length % 2 == 0):
    middle = (int) (length / 2)
    avg = (list[middle] + list[middle-1]) / 2
    return avg
```

```python
def Q1(list):
  list.sort()
  q = (int) (len(list) / 4)

  return list[q]
```

```python
def Q3(list):
  list.sort()

  q = (int) (3 * len(list) / 4)

  return list[q]
```

```python
def variance(list):
  result = 0
  mean = average(list)

  for n in list:
    result = result + (mean-n)**2

  return result / len(list)
```

```python
def standard_deviation(list):
  std = variance(list) ** 0.5
  return std
```
