```python
def sum(list):
  sum = 0
  for i in range(len(list)):
    sum = sum + list[i]
  
  return sum
```

```python
def max(list):
  result = list[0]
  for i in range(len(list)):
    if(list[i]>result):
      result = list[i]
  
  return result
```

```python
def min(list):
  result = list[0]
  for i in range(len(list)):
    if(list[i]<result):
      result = list[i]
  
  return result
```

```python
def median(list):
  list.sort()
  length = len(list)
  if(length%2==1):
    middle = (int) (length/2)
    return list[middle]

  if(length%2==0):
    middle = (int) (length/2)
    avg = (list[middle] + list[middle+1]) / 2
    return avg

```

```python
def mean(list):
  avg = sum(list)/len(list)
  return avg
```

```python
def Q1(list):
  list.sort()
  length = len(list)
  q = (int) (length/4)

  return list[q]
```

```python
def Q3(list):
  list.sort()
  length = len(list)
  q = (int) (3 * (length/4))

  return list[q]
```

```python
def variance(list):
  sum = 0;
  avg = mean(list)
  for i in range(len(list)):
    sum = sum + ((list[i] - avg) ** 2)

  variance = sum / len(list)

  return variance
```

```python
def standard_deviation(list):
  st = variance(list) ** 0.5
  return st
```

```python
list1 = [2,5,1,6,7,9,1,3,5,6]

print('Max is: ', max(list1))
print('Min is: ', min(list1))
print('Median is: ', median(list1))
print('Mean is: ', mean(list1))
print('Q1 is: ', Q1(list1))
print('Q3 is: ', Q3(list1))
print('Variance is: ', variance(list1))
print('Standard Deviation is: ', standard_deviation(list1))
```
